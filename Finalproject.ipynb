### 1. Continuous Forecasting using Logistic Growth Model (Sigmoid Curve)
We will use `scipy.optimize.curve_fit` to model the historical data of each province under a logistic growth formulation:
$$V(t) = \frac{L}{1 + e^{-k(t - t_0)}}$$
where $L$ is the carrying capacity, $k$ is the growth rate, and $t_0$ is the midpoint year.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Filter out total/sub-total rows
df_provinces = df[~df['Province'].str.strip().str.upper().isin(['TOTAL', 'SUB-TOTAL', 'SUBTOTAL'])].copy()
years = sorted([int(col.strip()) for col in df_provinces.columns if col.strip().isdigit()])
years_arr = np.array(years)

# Define the logistic growth function
def logistic_growth(t, L, k, t0):
    return L / (1 + np.exp(-k * (t - t0)))

future_years = list(range(2021, 2031))
forecast_results = []

for idx, row in df_provinces.iterrows():
    province = row['Province']
    y_values = pd.to_numeric(row[list(map(str, years))].values, errors='coerce')

    # Filter out NaNs for fitting
    mask = ~np.isnan(y_values)
    x_fit = years_arr[mask]
    y_fit = y_values[mask]

    if len(x_fit) < 3:
        continue

    # Initial parameter guesses: L = max value, k = 0.1, t0 = median year
    p0 = [max(y_fit) if max(y_fit) > 0 else 1.0, 0.1, np.median(x_fit)]

    try:
        # Fit logistic growth curve
        popt, _ = curve_fit(logistic_growth, x_fit, y_fit, p0=p0, maxfev=10000)
        predictions = logistic_growth(np.array(future_years), *popt)
    except Exception:
        # Fallback to linear trend if optimization does not converge
        coef = np.polyfit(x_fit, y_fit, 1)
        predictions = np.polyval(coef, np.array(future_years))

    res_dict = {'Province': province}
    for yr, pred in zip(future_years, predictions):
        res_dict[str(yr)] = max(0.0, round(pred, 2))
    forecast_results.append(res_dict)

df_logistic_forecast = pd.DataFrame(forecast_results)
print("Forecasted Continuous Values using Logistic Curve Fitting (2021-2030):")
display(df_logistic_forecast.head(10))
```

### 1. Trend Forecasting up to 2030 using Naive Bayes Classification
Because standard Naive Bayes classifiers work on categories, we formulate the forecasting problem by predicting the class variable: **"Will the province's value exceed its historical average in a given year?"** (1 for Yes, 0 for No). We use historical years and lagged performance indicators as features to predict these probabilities up to 2030.

```python
from sklearn.naive_bayes import GaussianNB
import numpy as np
import pandas as pd

# Exclude totals to keep only province-level data
df_provinces = df[~df['Province'].str.strip().str.upper().isin(['TOTAL', 'SUB-TOTAL', 'SUBTOTAL'])].copy()
years = sorted([int(col.strip()) for col in df_provinces.columns if col.strip().isdigit()])

# Melt dataframe for year-by-year analysis
df_melted = df_provinces.melt(id_vars=['Province'], value_vars=[str(y) for y in years], var_name='Year', value_name='Value')
df_melted['Year'] = pd.to_numeric(df_melted['Year'])
df_melted['Value'] = pd.to_numeric(df_melted['Value'], errors='coerce').fillna(0.0)

# Compute a baseline (mean value) for each province
baseline_map = df_melted.groupby('Province')['Value'].mean().to_dict()

# Classify targets: 1 if Value is above the historical average, 0 otherwise
df_melted['Target'] = df_melted.apply(lambda row: 1 if row['Value'] > baseline_map[row['Province']] else 0, axis=1)

# Train Gaussian Naive Bayes and forecast the probability of being above baseline up to 2030
nb_forecast = []
future_years = list(range(2021, 2031))

for province in df_provinces['Province'].unique():
    prov_data = df_melted[df_melted['Province'] == province]
    X = prov_data[['Year']].values
    y = prov_data['Target'].values

    # Gaussian Naive Bayes requires at least two classes to fit
    if len(np.unique(y)) < 2:
        # Defaulting constant probabilities if no historical state variation exists
        constant_prob = 1.0 if y[0] == 1 else 0.0
        prob_future = [constant_prob] * len(future_years)
    else:
        gnb = GaussianNB()
        gnb.fit(X, y)
        # Predict probability of class 1 (growth above baseline)
        prob_future = gnb.predict_proba(np.array(future_years).reshape(-1, 1))[:, 1]

    res_row = {'Province': province}
    for yr, p in zip(future_years, prob_future):
        res_row[f'{yr}_Growth_Prob'] = round(p, 4)
    nb_forecast.append(res_row)

df_nb_forecast = pd.DataFrame(nb_forecast)
print("Naive Bayes Probabilities of Values Exceeding Historical Averages (2021 to 2030):")
display(df_nb_forecast.head(10))
```

### 2. Visualizing Naive Bayes Growth Probabilities
Let's visualize the forecasted growth probabilities up to 2030 for some sample provinces.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sample_provinces = ['QUEZON CITY, (NCR   SECOND DISTRICT)', 'CAVITE', 'PAMPANGA', 'CEBU', 'ILOCOS NORTE']

plt.figure(figsize=(12, 6))
for province in sample_provinces:
    row = df_nb_forecast[df_nb_forecast['Province'] == province]
    if not row.empty:
        probs = [row[f'{yr}_Growth_Prob'].values[0] for yr in future_years]
        plt.plot(future_years, probs, marker='o', label=province)

plt.title('Naive Bayes Predicted Probability of Exceeding Historical Baseline (2021-2030)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### 1. Continuous Forecasting using Decision Tree Regression
We will use a `DecisionTreeRegressor` trained on historical data. For each province, we train the tree using the years as features and the historical values as targets, then make projections for years 2021 to 2030.

```python
from sklearn.tree import DecisionTreeRegressor
import numpy as np
import pandas as pd

# Filter out total/sub-total rows
df_provinces = df[~df['Province'].str.strip().str.upper().isin(['TOTAL', 'SUB-TOTAL', 'SUBTOTAL'])].copy()
years = sorted([int(col.strip()) for col in df_provinces.columns if col.strip().isdigit()])
years_arr = np.array(years)

future_years = list(range(2021, 2031))
dt_forecast_results = []

for idx, row in df_provinces.iterrows():
    province = row['Province']
    y_values = pd.to_numeric(row[list(map(str, years))].values, errors='coerce')

    # Filter out NaNs for training
    mask = ~np.isnan(y_values)
    X_train = years_arr[mask].reshape(-1, 1)
    y_train = y_values[mask]

    if len(X_train) == 0:
        continue

    # Initialize and fit Decision Tree Regressor
    # We set a moderate max_depth to avoid extreme overfitting on pure time step values
    dt_regressor = DecisionTreeRegressor(max_depth=5, random_state=42)
    dt_regressor.fit(X_train, y_train)

    # Predict values for 2021 to 2030
    predictions = dt_regressor.predict(np.array(future_years).reshape(-1, 1))

    res_dict = {'Province': province}
    for yr, pred in zip(future_years, predictions):
        res_dict[str(yr)] = max(0.0, round(pred, 2))
    dt_forecast_results.append(res_dict)

df_dt_forecast = pd.DataFrame(dt_forecast_results)
print("Forecasted Continuous Values using Decision Tree Regression (2021-2030):")
display(df_dt_forecast.head(10))
```

Let's plot both historical and predicted trends to see how the Decision Tree Regressor performs.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sample_provinces = ['QUEZON CITY, (NCR   SECOND DISTRICT)', 'CAVITE', 'PAMPANGA', 'CEBU', 'ILOCOS NORTE']

plt.figure(figsize=(15, 10))
colors = sns.color_palette("tab10", len(sample_provinces))

for province, color in zip(sample_provinces, colors):
    # Extract historical data
    prov_data = df_provinces[df_provinces['Province'] == province]
    if prov_data.empty:
        continue

    hist_y = pd.to_numeric(prov_data[list(map(str, years))].values[0], errors='coerce')
    plt.scatter(years, hist_y, color=color, alpha=0.6, label=f'{province} (Historical)')

    # Extract predicted data
    forecast_row = df_dt_forecast[df_dt_forecast['Province'] == province]
    if not forecast_row.empty:
        forecast_y = forecast_row[[str(y) for y in future_years]].values[0]
        full_x = list(years) + future_years
        full_y = list(hist_y) + list(forecast_y)

        plot_mask = ~np.isnan(full_y)
        plot_x = np.array(full_x)[plot_mask]
        plot_y = np.array(full_y)[plot_mask]

        plt.plot(plot_x, plot_y, color=color, linestyle='--', linewidth=2, label=f'{province} (Decision Tree Forecast)')

plt.title('Historical and Decision Tree Regression Forecasts (1988 - 2030)', fontsize=16)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Value', fontsize=12)
plt.axvline(x=2020, color='red', linestyle=':', label='Forecast Horizon Start (2020)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### 1. Continuous Forecasting using K-Nearest Neighbors (KNN) Regression

We will use a `KNeighborsRegressor` trained on historical year-value pairs for each province. KNN regression makes predictions by averaging the values of its $K$ nearest historical temporal neighbors.

```python
from sklearn.neighbors import KNeighborsRegressor
import numpy as np
import pandas as pd

# Filter out total/sub-total rows
df_provinces = df[~df['Province'].str.strip().str.upper().isin(['TOTAL', 'SUB-TOTAL', 'SUBTOTAL'])].copy()
years = sorted([int(col.strip()) for col in df_provinces.columns if col.strip().isdigit()])
years_arr = np.array(years)

future_years = list(range(2021, 2031))
knn_forecast_results = []

for idx, row in df_provinces.iterrows():
    province = row['Province']
    y_values = pd.to_numeric(row[list(map(str, years))].values, errors='coerce')

    # Filter out NaNs for training
    mask = ~np.isnan(y_values)
    X_train = years_arr[mask].reshape(-1, 1)
    y_train = y_values[mask]

    if len(X_train) == 0:
        continue

    # Initialize and fit KNN Regressor
    # Set n_neighbors dynamically based on available data size to avoid errors
    n_neighbors = min(3, len(X_train))
    if n_neighbors > 0:
        knn_regressor = KNeighborsRegressor(n_neighbors=n_neighbors, weights='distance')
        knn_regressor.fit(X_train, y_train)
        predictions = knn_regressor.predict(np.array(future_years).reshape(-1, 1))
    else:
        predictions = np.zeros(len(future_years))

    res_dict = {'Province': province}
    for yr, pred in zip(future_years, predictions):
        res_dict[str(yr)] = max(0.0, round(pred, 2))
    knn_forecast_results.append(res_dict)

df_knn_forecast = pd.DataFrame(knn_forecast_results)
print("Forecasted Continuous Values using KNN Regression (2021-2030):")
display(df_knn_forecast.head(10))
```
